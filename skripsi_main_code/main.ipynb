# 📝 DOKUMENTASI DEPTHESTIMATOR CLASS

## Overview

`DepthEstimator` adalah class yang mengemas semua fungsi depth estimation dalam satu interface yang mudah digunakan.

## Keuntungan Class-Based Approach

1. **State Management**: Model dan parameter disimpan sebagai instance variables
2. **Reusability**: Setup sekali, pakai berkali-kali tanpa pass parameter berulang
3. **Encapsulation**: Semua fungsi terkait dalam satu class
4. **Konsistensi**: Pola yang sama dengan YOLO (class-based)

## Methods Utama

### `__init__()`
Initialize DepthEstimator dengan konfigurasi awal.

### `load_camera_calibration()` (static method)
Load parameter kalibrasi dari file JSON.

### `estimate_depth(image)`
Estimasi depth map dari single image.

### `scale_recovery(depth_map)`
Konversi depth relatif ke absolut (meter).

### `visualize_depth(depth_map)`
Visualisasi depth map dengan colormap.

### `create_depth_overlay(image, depth_colored)`
Overlay depth visualization di atas image.

### `process_frame(image_path)`
Pipeline lengkap untuk single image.

### `process_video(video_path)`
Pipeline lengkap untuk video per-frame.
